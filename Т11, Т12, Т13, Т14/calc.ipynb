```python
# Файл для построение графиков и расчетов.

from scipy.stats import f, t
import matplotlib.pyplot as plt
import numpy as np

plt.style.use("ggplot")
figure, axis = plt.subplots(figsize=(10, 6))
axis: plt.Axes


theta_1 = np.linspace(0, 2.5, 1000)

c1 = 0.7673
c2 = 1.2716

w_1 = f(138, 999).cdf(c1 * theta_1) + 1 - f(138, 999).cdf(c2 * theta_1)

axis.plot(theta_1, w_1, color='blue')
axis.set_title("Мощность критерия в зависимости от θ = b²/a² (T13)")
axis.set_xlabel("θ = b²/a²")
axis.set_ylabel("W")
axis.set_xlim(theta_1.min(), theta_1.max())
```

```python
x = np.array([-1.11, -6.10, 2.42])
y = np.array([-2.29, -2.91])

n_x = len(x)
n_y = len(y)

s_x = np.sum((x - np.mean(x))**2) / (n_x - 1)
s_y = np.sum((y - np.mean(y))**2) / (n_y - 1)


d = (np.mean(x) - np.mean(y)) / (np.sqrt((7 * (s_x + s_y)) / 18))
print(d)
```

```python
figure, axis = plt.subplots(figsize=(10, 6))
axis: plt.Axes


theta_2 = np.linspace(0, 25, 1000)
c = t(3).ppf(0.95)
coeff = np.sqrt((7 * (s_x + s_y)) / 18)

w_2 = 1 - t(3).cdf(c - (theta_2 / coeff))

axis.plot(theta_2, w_2, color='blue')
axis.set_title("Мощность критерия в зависимости от θ = a - b (T14)")
axis.set_xlabel("θ = a - b")
axis.set_ylabel("W")
axis.set_xlim(theta_2.min())
```
